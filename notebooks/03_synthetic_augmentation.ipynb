# 03 · Synthetic augmentation (PCA-space jitter)
Generated per disorder, inside training folds only.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np, matplotlib.pyplot as plt
%matplotlib inline
from src.augmentation import pca_jitter
from sklearn.decomposition import PCA
```

```python
fm = pd.read_csv('../data/processed/feature_matrix.csv', index_col=0)
tr = fm[fm.split=='train']
X = tr.drop(columns=['split','disorder']).reset_index(drop=True); g = tr.disorder.values; y = (g!='none').astype(int)
```

```python
Xa, ya = pca_jitter(X, y, g, n_new_per_pos=5, alpha=0.15)
print(len(X), '->', len(Xa), 'positives', y.sum(), '->', ya.sum())
```

```python
p = PCA(2).fit(X.values); P = p.transform(X.values); Pa = p.transform(Xa.values[len(X):])
plt.scatter(*P[y==0].T, s=2, c='lightgrey'); plt.scatter(*Pa.T, s=5, alpha=.4); plt.scatter(*P[y==1].T, s=12, c='r');
```

**Leakage warning:** augmenting before the train/test split puts near-copies of test patients into training.
